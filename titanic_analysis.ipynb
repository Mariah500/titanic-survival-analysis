```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from tabulate import tabulate
train=pd.read_csv('train.csv')
train.dropna()
#inspect the data(shape,types,missing values
print(train.info())#shows shape,data types and non null counts in one print out
print(train.shape)#shows data rows and columns
print(train.dtypes)#shows each column's data type
print(train.isnull().sum())#gives a count of missing and incomplete data per column
#survival rates by gender, class and age group
print(train.groupby('Sex')['Survived'].mean())
print(train.groupby('Pclass')['Survived'].mean())
train['AgeGroup']=pd.cut(train['Age'],bins=[0,12,18,35,60,100],labels=['Child','Teen','Adult','Midage','Senior'])
print(train.groupby('AgeGroup')['Survived'].mean())

```
